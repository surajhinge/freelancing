```python
import pandas as pd

# 📌 Enter your Excel file path here
excel_path = "D:/freelancing_work/github_repo/freelancing/refactored_NSE_Automate/algo_trading/Filtered_Instruments.xlsx"

try:
    # Read the Excel file
    df = pd.read_excel(excel_path)

    print("✅ Excel file loaded successfully!")
    print(f"ℹ Total records: {len(df)}")

    if "Date" in df.columns:
        print("\n📅 Dates (only date, no time):")
        
        # Convert to date only
        df["Date"] = pd.to_datetime(df["Date"]).dt.date
        
        for date in df["Date"]:
            print(date)

    else:
        print("\n⚠️ 'Date' column not found! Columns present:", df.columns.tolist())

except FileNotFoundError:
    print(f"❌ File not found: {excel_path}")
except Exception as e:
    print(f"❌ Error: {e}")

```
